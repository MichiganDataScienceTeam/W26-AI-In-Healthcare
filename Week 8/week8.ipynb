### 3. Widgets

```python
# creates a slider for setting bmi
bmi = st.slider("Select your bmi", min_value=18, max_value=25, value=22)
st.write("Your bmi is:", bmi)

# Creates a dropdown box
smokes = st.selectbox("Do you smoke?", ["Yes", "No", "Sometimes"])
st.write("Smoking Status:", smokes)

# Creates a button and a rule for clicking it
if st.button("Click please"):
    st.write("Clicked")

# there are a lot of other widgets on the docs that might be useful as well
```

### 4. Dataframes

```python
# sample dataframe
df = pd.DataFrame({
    'Age': [20, 24, 65, 87, 43],
    'BMI': [19, 18.5, 20, 25, 24.4],
    'Diabetes Risk': [0.5, 0.25, 0.75, 1.0, 2.0]
})

# display the dataframe
st.dataframe(df)

# custom height
st.dataframe(df, height=200)

# custom width
st.dataframe(df, width=200)

# make dataframe take up whole container width
st.dataframe(df, use_container_width=True)

# Hide index
st.dataframe(df, hide_index=True)
```

### 5. Visualizations

```python
# creates a pairplot of the dataframe
fig = plt.figure(figsize=(15, 5))
sns.pairplot(df)

# plots the pairplot using st.pyplot()
st.pyplot(fig, clear_figure=True, width="stretch")

# you can also use plt.gcf() which gets the current figure if you don't want to do plt.figure() at the beginning

# the following line can be used to create a separating line between your visualizations
st.divider()
```

### 6. Model Integration

```python
# load your model
model = joblib.load("./model.pkl")

# We can create input feature slides/selection boxes for your model using st.form()
with st.form("input features"):

    # numerical features
    age = st.slider("Age", 18, 100, 30)
    bmi = st.slider("BMI", 18, 25, 22)

    # categorical feature
    smokes = st.selectbox("Smokes", ["Yes", "No", "Occasionally"])

    # button for running the prediction
    submitted = st.form_submit_button("Run prediction")

if submitted:

    # encodes categorical variable
    smokes_mapping = {"Yes": 1, "No": 0, "Occasionally": 2}
    smokes_encoded = smokes_mapping[smokes]

    # create a feature numpy array (make sure it's 2d with a single column and the order of the features matches up with how the model was trained)
    features = np.array([[age, bmi, smokes_encoded]])

    # run predictions
    prediction = model.predict(features)

    # we can also get a predicted probability for classification
    prob = model.predict_proba(features)[0]

    # display the results
    st.write("Prediction:", prediction)
    st.write("Predicted Probability:", prob)
```
